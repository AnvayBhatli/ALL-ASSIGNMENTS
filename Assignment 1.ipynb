```python
SUBJECTS = ("Maths", "Physics", "Programming")

students = {
    101: ("Aarav Sharma", "Computer Science", [85, 90, 78]),
    102: ("Priya Patil", "Electronics", [72, 68, 80]),
    103: ("Rohan Verma", "Mechanical", [65, 70, 60]),
}


def display_students(title="Student Records"):
    print(f"\n{title}")
    print("-" * 84)
    header = f"{'Roll':<6}{'Name':<18}{'Branch':<24}"
    header += "".join(f"{s:<12}" for s in SUBJECTS) + "Total"
    print(header)
    print("-" * 84)
    if not students:
        print("No records found.")
    for roll, (name, branch, marks) in sorted(students.items()):
        row = f"{roll:<6}{name:<18}{branch:<24}"
        row += "".join(f"{m:<12}" for m in marks) + str(sum(marks))
        print(row)
    print("-" * 84)


def add_student(roll, name, branch, marks):
    """1. Add a new student record."""
    if roll in students:
        print(f"Roll number {roll} already exists. Use update instead.")
        return
    students[roll] = (name, branch, list(marks))   # tuple holding a list
    print(f"Student {roll} ({name}) added.")


def delete_student(roll):
    """2. Delete an existing student record."""
    if roll in students:
        removed = students.pop(roll)
        print(f"Student {roll} ({removed[0]}) deleted.")
    else:
        print(f"Roll number {roll} not found.")


def update_student(roll, name=None, branch=None, marks=None):
    """3. Update details of a student."""
    if roll not in students:
        print(f"Roll number {roll} not found.")
        return
    old_name, old_branch, old_marks = students[roll]
    students[roll] = (
        name if name is not None else old_name,
        branch if branch is not None else old_branch,
        list(marks) if marks is not None else old_marks,
    )
    print(f"Student {roll} updated.")


def main():
    display_students("Initial Student Records")
    
    add_student(104, "Sneha Kulkarni", "Information Technology", [88, 92, 95])

    delete_student(102)

    update_student(103, branch="Civil", marks=[70, 75, 68])
    update_student(101, name="Aarav S. Sharma")

    display_students("Final Student Records")


if __name__ == "__main__":
    main()
```
