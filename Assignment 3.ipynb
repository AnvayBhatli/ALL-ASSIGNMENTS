```python
"""Q.Check whether a triangle with given side lengths is right-angled."""

import math


def is_valid_triangle(a, b, c):
    """Sides must be positive and satisfy the triangle inequality."""
    return a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a


def is_right_angled(a, b, c):
    x, y, z = sorted((a, b, c))          # z is the longest side
    return math.isclose(x ** 2 + y ** 2, z ** 2, rel_tol=1e-9)


def main():
    a = float(input("Enter length of side 1: "))
    b = float(input("Enter length of side 2: "))
    c = float(input("Enter length of side 3: "))

    if not is_valid_triangle(a, b, c):
        print("These sides do not form a valid triangle.")
    elif is_right_angled(a, b, c):
        print("The triangle is a right-angled triangle.")
    else:
        print("The triangle is NOT a right-angled triangle.")


if __name__ == "__main__":
    main()
```
